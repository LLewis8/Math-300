```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
```

### Problem 1
Suppose 

$$\frac{dy}{dt}=1.2y.$$

instead of .5 from lecture. Solve again.

How does the graph change?



```python
def model_1(t, y):
    return 1.2 * y

t_span = (0, 5)
t_values = np.linspace(0, 5, 200)
sol_1 = solve_ivp(model_1, t_span, y0=[1], t_eval=t_values)

plt.figure(figsize=(10, 6))
plt.plot(sol_1.t, sol_1.y[0], label='Model 1: dy/dt = 1.2y', color='blue')
plt.title('Solution of the Differential Equation')
plt.xlabel('Time (t)')
plt.ylabel('y(t)')
plt.grid()
plt.show()
```

```python
#The growth rate of this graph rises much more quickly then the one shown in class because the growth rate is increased from 0.5 to 1.2.
```

## Problem 4

Modify the logistic model by ranging $K$ from 0 to 50 with 100 slices.

How does the solution change?


```python
def logistic_model(t, y, r, K):
    return r * y * (1 - y / K)

r = 0.5
initial_pop = 1

t_span = (0, 20)
t_values = np.linspace(0, 20, 300)

k_values = np.linspace(0.5, 50, 100)

plt.figure(figsize=(10, 6))
for K in k_values:
    solution = solve_ivp(logistic_model, t_span, y0=[initial_pop], t_eval=t_values, args=(r, K))
    
    plt.plot(solution.t, solution.y[0], label=f'K={K:.1f}', alpha=0.5)

plt.xlabel('Time (t)')
plt.ylabel('Population (y)')
plt.title('Logistic Growth Model for Different Carrying Capacities')
plt.grid()
plt.show()
```

```python
#The population approaches the carrying capacity K as time progresses. As K increases, the solution levels off at a higher population.
#Small values of K result in lower limiting populations, while larger values of K allow for higher populations. The growth rate r determines how quickly the population approaches the carrying capacity.
```

## Problem 5

Create your own model.

Ideas include

- Predator-prey model
- Cooling of coffee
- Disease spread
- Population growth
- Bank account with continuous interest

Model the system using a differential equation and solve it with `solve_ivp()`

```python
room_temp = 70
initial_temp = 190
cooling_rate = 0.08

def coffee_cooling(t, T):
    return -cooling_rate * (T - room_temp)  

t_span = (0, 60)
t_values = np.linspace(0, 60, 300)
solution = solve_ivp(coffee_cooling, t_span, y0=[initial_temp], t_eval=t_values)

plt.figure(figsize=(10, 6))
plt.plot(solution.t, solution.y[0], label='Coffee Cooling', color='brown')
plt.axhline(y=room_temp, color='blue', linestyle='--', label='Room Temperature')
plt.title('Cooling of Coffee Over Time')
plt.xlabel('Time (minutes)')
plt.ylabel('Temperature (°F)')
plt.legend()
plt.grid()
plt.show()  
```
