### Problem 4

The derivative of a function \(f(x)\) at a point \(x\) is defined as the limit

$
f'(x)=\lim_{h\to0}\frac{f(x+h)-f(x)}{h}
$

provided the limit exists.

Define a function $g(h)$ that computes the quotient above for a given function $f$ near a point $x$. Then, vectorize $g$. 


Using `np.linspace()`, let $f(x)=sin(x)$ and compute $g$ at $10^{-1}, 10^{-2}, 10^{-3}, 10^{-4}$. This will be an estimate for the derivative of $f$. Compare to the actual derivative.

````python
import math
````

````python
import numpy as np

x = np.pi

def f(x):
    return np.sin(x)

def g(h):
    return (f(x + h) - f(x)) / h

G = np.vectorize(g)
h_values = 10.0 ** np.linspace(-1, -4, 4)
estimated_dv = G(h_values)
actual_dv = np.cos(x)

print('h values:', h_values)
print('Estimated derivatives:', estimated_dv)
print('Actual derivative:', actual_dv)
````

###  Problem 5
Let

$
A=
\begin{bmatrix}
2 & 1 \\
3 & 4
\end{bmatrix},
\qquad
B=
\begin{bmatrix}
5 & 2 \\
1 & 6
\end{bmatrix}
$

Compute the product.

````python
a = (
    [2, 1],
    [3, 4]
)
b = (
    [5, 2],
    [1, 6]
)
A = np.array(a).reshape(2,2)
B = np.array(b).reshape(2,2)

product = A @ B
print(product)
````
