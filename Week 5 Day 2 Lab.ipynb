### Problem 1
Suppose $f(x)=x^2+6x+10$

1. Plot the function to visualize the minimum
2. Find the minimum using `scipy`

````python
import scipy as sp
import matplotlib.pyplot as plt
import numpy as np
````

````python
x = np.linspace(-10,10,500)
plt.plot(x, x ** 2 + 6 * x + 10)

plt.grid()
plt.show()
````

````python
from scipy.optimize import minimize
````

````python
x0 = 0
def f(x):
    return x ** 2 + 6 * x + 10
result = minimize(f, x0=0)

print(result)

# Minimum occurs at x = -3
f(-3)
````


## Problem 2

Find the maximum of $50x-x^2$.

````python
def g(x):
    return -(50 * x - x ** 2)

maximum = minimize(g, x0=0)

print(maximum)
print('Maximum x = ', maximum.x)
print('Maximum value = ', -maximum.fun)
````

## Problem 3

Minimize

```python
np.sin(x)+x**2
```

using different starting values.

Do you obtain the same solution? Compute the derivative using `sympy`. When is this derivative zero?

````python
import sympy as sp

def h(x):
    return np.sin(x) + x ** 2

starts = [-5, -2, 0, 2, 5]
for s in starts:
    result = minimize(h, x0=s)

    print('Start = ', s)
    print(result.x)
    print(result.fun)
    print()

x = sp.symbols('x') 
function = sp.sin(x) + x ** 2

derivative = sp.diff(function, x)

print(derivative)

critical = sp.nsolve(derivative, -1)

print(critical)
````

### Problem 4
Design a function with multiple minima and optimize.

````python
def my_function(x):
    return np.sin(3 * x) + x ** 2

x = np.linspace(-4, 4, 500)

plt.plot(x, my_function(x))
plt.grid()
plt.show()

for start in [-3, -1, 1, 3]:
    result = minimize(my_function, x0=start)

    print('Start = ', start)
    print('Minimum x = ', result.x)
    print('Minimum value = ', result.fun)
    print()
````
