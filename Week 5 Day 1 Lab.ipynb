### Problem 1
Write your own function performs vectors addition on lists instead of arrays.

```python
import numpy as np

def vector_add(v1, v2):
    result = []

    for i in range(len(v1)):
        result.append(v1[i] + v2[i])
    return result

a = [1, 2, 3]
b = [4, 5, 6]
print(vector_add(a, b))
```

### Problem 3
Make a 2 by 2 invertible matrix, A.
1. Compute its eigenvectors and plot them using matplotlib
2. Compute A times the eigenvectors and plot them on the same plot.
3. What do you notice?

```python
import matplotlib.pyplot as plt

A = np.array([[3, 1], [1, 2]])

values, vectors = np.linalg.eig(A)
plt.figure(figsize=(6, 6))
origin = [0, 0]

for i in range(2):
    v = vectors[:,i]

    plt.quiver(origin[0], origin[1], v[0], v[1], angles='xy', scale_units='xy',
              scale=1, label=f'Eigenvector {i +1}' 
        )
    
    Av = A @ v
    plt.quiver(origin[0], origin[1], Av[0], Av[1], angles='xy', scale_units='xy',
              scale=1, label=f'A*v{i +1}' 
        )

plt.xlim(-4, 4)
plt.ylim(-4, 4)
plt.grid()
plt.legend()
plt.show()

print('Eigenvalues:', values)
print('\nEigenvectors:', vectors)
```

```python
# Multiplying the eigenvector by the matrix changes only its length. The transformed vector stays on the same line as the original eigenvector.
```
